```python
filterd_list = [(f, f.split("/")[-1]) for f in tdms_list if 'vernier' in f.lower()]

# filterd_list = [f for f in filterd_list if 'k1' in f]

filterd_list = [f[0] for f in filterd_list if f[1] in vaild_file]

filterd_list
```

```python
data_list = []

for tdms_file_path in filterd_list:
    file_name = tdms_file_path.split("/")[-1].replace('__', '_')
    file_name = file_name.replace('.tdms', '').split('_')
    file_name = ''.join([file_name[0], '_', file_name[2], '_', file_name[3]])
    
    tdms_file = TdmsFile.read(tdms_file_path)
    
    for group in tdms_file.groups():
        for channel in group.channels():
            data_len = np.shape(channel.data)[0]
            if (file_name[0] != 'z' and channel.name == 'Voltage'):
                data_list.append({
                    'excitation': channel.data,
                    'identifier': file_name  
                })
    print(file_name)
```

```python
df_magnitude = pd.DataFrame(data_list)
```

```python
df_magnitude['len'] = df_magnitude['excitation'].apply(len)

df_magnitude['excitation'] = df_magnitude['excitation'].apply(lambda x: x[50000:100000])
df_magnitude['len'] = df_magnitude['excitation'].apply(len)
```

```python
df_magnitude = df_magnitude.drop(['len'], axis=1)

df_magnitude
```

```python
def group_help(x: str) -> str:
    x = x.split('_')
    
    return ''.join([x[0], '_', x[2]])
    

df_magnitude['identifier'] = df_magnitude['identifier'].apply((lambda x: group_help(x)))
```

```python
#def stack_signal(x: list) -> np.array:
    

df_magnitude = df_magnitude.groupby('identifier').apply(lambda x: np.stack(x['excitation']))
```

```python
df_magnitude = pd.DataFrame(df_magnitude)
```

```python
df_magnitude = df_magnitude.reset_index()
df_magnitude.columns = "id", 'signals'
```

```python
tmp_folder = r'../tmp'
program_path = r'../build/CUDA'
paths = [r'../tmp/sig_0.txt', r'../tmp/sig_1.txt', r'../tmp/sig_2.txt']

# Upewnij się, że folder istnieje
os.makedirs(tmp_folder, exist_ok=True)

results = []

for row_idx, row in tqdm(df_magnitude.iterrows(), total=df_magnitude.shape[0]):
    signals = row['signals']

    for sig_idx in range(3):
        sig_data = np.array(signals[sig_idx])
        np.savetxt(paths[sig_idx], sig_data, fmt='%.6f')
    
    # 2. Odpalamy proces (teraz mamy pewność, że pliki istnieją i są 3)
    try:
        process = subprocess.run(
            [program_path, paths[0], paths[1], paths[2]],
            stdout=subprocess.PIPE,  # Zamiennik capture_output=True
            stderr=subprocess.PIPE,  # Zamiennik capture_output=True
            universal_newlines=True, # Zamiennik text=True w Pythonie 3.6
            check=True
        )

        # Wynik odbierasz tak samo
        prediction = process.stdout.strip()
        print(prediction)
        results.append(prediction)

    except subprocess.CalledProcessError as e:
        print(f"Błąd w wierszu {row_idx}: {e.stderr}")
        results.append("ERROR")

# 3. Na koniec dodaj wyniki do nowej kolumny
df_magnitude['prediction'] = results
```

```python
df_magnitude['prediction'] = df_magnitude['prediction'].apply(lambda x: x.split('\n')[-1])

df_magnitude['prediction'] = df_magnitude['prediction'].apply(lambda x: x.split())
```
